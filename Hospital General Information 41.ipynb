```python
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

for col in df.select_dtypes(include=['number']).columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)
IQR = Q3 - Q1

outliers = ((df.select_dtypes(include='number') < (Q1 - 1.5 * IQR)) |
            (df.select_dtypes(include='number') > (Q3 + 1.5 * IQR)))

print(outliers.sum())
```

<h3 style="

margin:0;">
📄Remove Outliers ¶

</h3>

</div>

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[~((df.select_dtypes(include='number') < lower_limit) |
          (df.select_dtypes(include='number') > upper_limit)).any(axis=1)]

print(df.shape)
```

* <b>Key Takeaway:</b> Private non-profit healthcare institutions deliver the highest volume of superior star ratings (4 & 5 stars), but severe reporting gaps across public and defense sectors mask the overall quality landscape[cite: 1, 2].

```python
# Bivariate Cross-tabulation table
type_emergency_ct = pd.crosstab(
    df['Hospital Type'], 
    df['Emergency Services'], 
    margins=True, 
    margins_name="Total"
)

print("--- Emergency Services by Hospital Type ---")
print(type_emergency_ct)

# Row-wise Percentage
print("\n--- Percentage Within Hospital Type (%) ---")
print((pd.crosstab(df['Hospital Type'], df['Emergency Services'], normalize='index') * 100).round(2))
```

```python
# Rating numeric aakki clean cheyyunnu
df_rate = df.copy()
df_rate['Hospital overall rating'] = pd.to_numeric(df_rate['Hospital overall rating'], errors='coerce')

# Bivariate Groupby Aggregation (Count, Mean, Median, Min, Max)
rating_by_type = df_rate.groupby('Hospital Type')['Hospital overall rating'].agg(
    Total_Hospitals='count',
    Average_Rating='mean',
    Median_Rating='median',
    Min_Rating='min',
    Max_Rating='max'
).round(2)

print("--- Hospital Overall Rating Summary by Hospital Type ---")
print(rating_by_type)
```

```python
# Rating numeric aakki clean cheyyunnu
df_rating = df.copy()
df_rating['Hospital overall rating'] = pd.to_numeric(df_rating['Hospital overall rating'], errors='coerce')
df_rating = df_rating.dropna(subset=['Hospital overall rating', 'Hospital Type'])

# Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_rating, x='Hospital Type', y='Hospital overall rating', palette='mako')

plt.title('Overall Rating Distribution across Hospital Types')
plt.xlabel('Hospital Type')
plt.ylabel('Star Rating (1 - 5)')
plt.xticks(rotation=15)
plt.show()
```

```python
# Rating numeric aakki clean cheyyunnu
df_multi = df.copy()
df_multi['Hospital overall rating'] = pd.to_numeric(df_multi['Hospital overall rating'], errors='coerce')

# Missing values remove cheyyunnu
df_multi_clean = df_multi.dropna(subset=['Hospital Type', 'Emergency Services', 'Hospital overall rating'])

# Visualization: Grouped Bar Chart (Average Rating by Hospital Type & Emergency Services)
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_multi_clean,
    x='Hospital Type',
    y='Hospital overall rating',
    hue='Emergency Services',
    palette='Set2',
    ci=None
)

plt.title('Average Hospital Overall Rating by Hospital Type & Emergency Services', fontsize=13, fontweight='bold')
plt.xlabel('Hospital Type', fontsize=11)
plt.ylabel('Average Star Rating (1 - 5)', fontsize=11)
plt.xticks(rotation=15)
plt.ylim(0, 5)
plt.legend(title='Emergency Services')
plt.tight_layout()
plt.show()
```
