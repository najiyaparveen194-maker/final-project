```python
df = pd.read_csv('Hospital_General_Information.csv')

count_columns = [
    'Count of Facility MORT Measures', 
    'Count of Facility Safety Measures', 
    'Count of Facility READM Measures', 
    'Count of Facility Pt Exp Measures', 
    'Count of Facility TE Measures'
]

for col in count_columns:

    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df[count_columns].dtypes)
```

```python
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

for col in df.select_dtypes(include=['number']).columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)
IQR = Q3 - Q1

outliers = ((df.select_dtypes(include='number') < (Q1 - 1.5 * IQR)) |
            (df.select_dtypes(include='number') > (Q3 + 1.5 * IQR)))

print(outliers.sum())
```

<h3 style="

margin:0;">
📄Remove Outliers ¶

</h3>

</div>

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[~((df.select_dtypes(include='number') < lower_limit) |
          (df.select_dtypes(include='number') > upper_limit)).any(axis=1)]

print(df.shape)
```

* <b>Acute Care Dominance:</b> Over 95% of <b>Acute Care Hospitals(2658)</b> and nearly all <b>Critical Access Hospitals(988)</b> offer emergency services, as immediate urgent care is central to their primary healthcare mission.
<br>
* <b>Specialized Facility Differences:</b> Specialized facilities or select administrative categories account for the small subset without emergency services, focusing instead on elective treatments, specialized rehabilitation, or outpatient procedures.
</div>

<div style="background:#F8BBD0; color: white; padding: 10px 15px; border-radius: 10px; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
    <h3 style="margin: 0; color: #FFFFFF; font-weight: 600; font-family: sans-serif;">8. 📈 How do Hospital Overall Ratings vary across Hospital Types?</h3>
</div>

```python
df_rate = df.copy()
df_rate['Hospital overall rating'] = pd.to_numeric(df_rate['Hospital overall rating'], errors='coerce')

rating_by_type = df_rate.groupby('Hospital Type')['Hospital overall rating'].agg(
    Total_Hospitals='count',
    Average_Rating='mean',
    Median_Rating='median',
    Min_Rating='min',
    Max_Rating='max'
).round(2)

print("--- Hospital Overall Rating Summary by Hospital Type ---")
print(rating_by_type)
```

```python
df_rating = df.copy()
df_rating['Hospital overall rating'] = pd.to_numeric(df_rating['Hospital overall rating'], errors='coerce')
df_rating = df_rating.dropna(subset=['Hospital overall rating', 'Hospital Type'])

# Boxplot
plt.figure(figsize=(10, 5))
sns.barplot(data=df_rating, x='Hospital Type', y='Hospital overall rating', palette='mako')

plt.title('Overall Rating Distribution across Hospital Types')
plt.xlabel('Hospital Type')
plt.ylabel('Star Rating (1 - 5)')
plt.xticks(rotation=15)
plt.show()
```
